```python
## SHRNUTÍ PŘEDEŠLÉHO ##

# Načteme tabulky o potravinách
food_sample_100 = pd.read_csv("food_sample_100.csv")
food_other = pd.read_csv("food_other.csv")
# Sjednotíme tabulky o potravinách do jedné tabulky food
food = pd.concat([food_sample_100, food_other], ignore_index=True)
# Načteme data o výživných látkách
food_nutrient = pd.read_csv("food_nutrient.csv")
# Propojíme tabulky o potravinách a výživných látkách
food_merged = pd.merge(food, food_nutrient, on="fdc_id")
# Načteme tabulku o značkách potravin
branded_food = pd.read_csv("branded_food.csv")
# Propojíme tabulku o značkách potravin a tabulkou o potravinách a výživných látkách
food_merged_brands = pd.merge(food_merged, branded_food, right_on="fdc_id", left_on="fdc_id", how="left")
# Přejmenujeme sloupec name na nutrient_name
food_merged_brands = food_merged_brands.rename(columns={"name": "nutrient_name"})
# NEPOVINNÝ KROK: Odstraníme nepotřebné sloupce footnote a min_year_acquired
food_merged_brands = food_merged_brands.drop(columns=["footnote", "min_year_acquired"])
```

```python
# Agregace a řazení-Agregace je operace, která "sloučí" více řádků tabulky do jednoho.U agregace většinou slučujeme řádky podle nějakého konkrétního sloupce.

# Najít potraviny s Proteiny --> PODMÍNKA
food_merged_brands_protein = food_merged_brands [food_merged_brands["nutrient_name"] == "Protein"]
# food_merged_brands_protein.info ()

# Rozdělit na kategorii ("branded_food_category") --> seskupení, GROUP BY V SQL
food_merged_brands_protein.groupby ("branded_food_category")

# Spočítat průměr (MEAN) za každou kategorii --> vyber "amount", počítej průměr (V PYTHON JE PRŮMĚR MEAN !!!POZOR!!!)
# food_merged_brands_protein.head() -- zobrazení tabulky,abych si našla podle čeho budu počítat průměr tj. podle amount 
food_merged_brands_protein.groupby ("branded_food_category") ["amount"].mean ()

# Seřadit pro přehlednost --> METODA SORT_VALUE()

# Řazení pomocí ASC(ascending),#.sort_values(ascending=False) zajistí řazení od největšího
# !!! POZOR !!! DESCENDING NEEXISTUJE ,řadíme tedy pomocí True/False !!!, head zajistí TOP 10 místo 193 řádků 

food_merged_brands_protein.groupby ("branded_food_category") ["amount"].mean().sort_values(ascending=False).head(10) 

# můžu napsat i tímto způsobem,vytvořit si proměnnou data_agg

data_agg = food_merged_brands_protein.groupby ("branded_food_category") ["amount"].mean()
data_agg.sort_values(ascending=False).head(10) 
```

```python
# CVIČENÍ Lipidy a Tuky
# Vytvoř novou tabulku food_merged_brands_lipid, do které pomocí dotazu vlož pouze řádky, které mají jako název výživné látky hodnotu Total lipid (fat)
food_merged_brands_lipid = food_merged_brands[
    food_merged_brands["nutrient_name"] == "Total lipid (fat)"
]
food_merged_brands_lipid.head(3)

# Proveď agregaci podle návu kategorie a seřaď výslednou tabulku tak, aby nahoře byly vidět kategorie s největším počtem tuků.
food_merged_brands_lipid_agg = food_merged_brands_lipid.groupby(
    "branded_food_category"
)["amount"].mean()

food_merged_brands_lipid_agg.sort_values(ascending=False)
```

```python
# CVIČENÍ Výrobci a kategorie
# Jaký výrobce produkuje jaké typy potravin.
# Proveď agregaci tabulky food_merged_brands podle dvou sloupců: brand_owner a branded_food_category.
# Sloupce musíš metodě food_merged_brands zadat jako seznam, tj. musíš použít hranaté závorky.
# Vyber sloupec fdc_id pro provedení agregace a použij agregaci nunique(), který vrátí počet unikátních hodnot.
# Nakonec použij metodu sort_values s tím, že chceš data seřadit sestupně.

food_merged_brands.groupby(
    ["brand_owner", "branded_food_category"]
)["fdc_id"].nunique().sort_values(ascending=False)
```

```python
# CVIČENÍ Hlasy
import pandas as pd

data = pd.read_csv('votes.csv')
# data
# data.info()

#Celkový počet odevzdaných hlasů v jednotlivých letech.
data.groupby('year')['candidatevotes'].sum()

#Celkový počet odevzdaných hlasů pro kandidáty a kandidátky jednotlivých stran pro volby v roce 2020.
data[data['year']==2020].groupby('party_detailed')['candidatevotes'].sum().sort_values(ascending=False)

# Celkový počet odevzdaných hlasů pro kandidáty dvou hlavních stran (DEMOCRAT a REPUBLICAN) pro všechny volby od roku 1980.
year = 1980
filter = (data['party_detailed'].isin(['DEMOCRAT','REPUBLICAN'])) &  (data['year']>=year)
data[filter].groupby('year')['candidatevotes'].sum().sort_values(ascending=False)

#Celkový počet odevzdaných hlasů pro stát MONTANA pro všechny roky.
data[data['state']=='MONTANA']['candidatevotes'].sum()
```

```python
# Použij metodu concat a spoj všechny tučňáky do jednoho dataframu. Kolik má nový dataframe řádek? 344 

penguins_adelie = pd.read_csv ("penguins_adelie.csv")
penguins_gentoo = pd.read_csv ("penguins_gentoo.csv")
penguins_chinstrap = pd.read_csv ("penguins_chinstrap.csv")

penguins = pd.concat ([penguins_adelie, penguins_gentoo, penguins_chinstrap], ignore_index=True)
penguins.info()
```

```python
# zkopírováno z kvízu 

island_data = {
    'Island': ['Torgersen', 'Biscoe', 'Dream'],
    'Latitude': [-64.7667, -66.0000, -64.7333],
    'Longitude': [-64.0667, -66.0000, -64.2333],
    'Location': [
        'Torgersen Island, near Palmer Station on the Antarctic Peninsula, close to Anvers Island',
        'Biscoe Islands, an archipelago off the west coast of the Antarctic Peninsula, south of the Palmer Archipelago',
        'Dream Island, part of the Argentine Islands, located near the western side of the Antarctic Peninsula'
    ]
} 

penguin_islands_df = pd.DataFrame(island_data) 
```

```python
# 1. Propojení tabulek, Pomocí metody merge připoj data o ostrovech k hlavnímu dataframu

merged_df = penguins.merge(penguin_islands_df, on='Island')
merged_df

# 2 Jaká je průměrná hodnota Latitude ve spojeném dataframu?
prumer = merged_df["Latitude"].mean()
prumer
```

```python
penguins = pd.read_csv ("penguins.txt")
penguins
```

```python
# Existují v datasetu chybějící hodnoty? Pokud ano, ve kterých sloupcích?
# Můžeš použít funkci .isnull() na rozpoznávání chybějících hodnot a pak sečíst jejich výskyt .sum()

penguins.isnull()
penguins.isnull().sum()
```

```python
# Jaké jsou průměrné hodnoty mean() hmotnosti tučňáků podle jejich druhu? Která odpověď je správně?

# Seskupíme data podle sloupce 'species', Seskupení dat (groupby)
seskupeno = penguins.groupby('species')

# Výběr sloupce a výpočet průměru (mean)
prumer = penguins.groupby('species')['body_mass_g'].mean()
prumer
```

```python
# Který druh tučňáka má největší maximální hmotnost a jaká je tato hmotnost?
penguins.groupby("species")["body_mass_g"].max().sort_values(ascending=False) # ascending=False --> řadí data sestupně (od největšího)
```

```python
# Jaký je počet tučňáků podle jejich pohlaví v každém druhu? 
penguins.groupby(["species", "sex"]).size().sort_values(ascending=False)
```

```python
# Jaké jsou minimální a maximální hodnoty délky křídel tučňáků a jak se liší mezi druhy? 
# Který druh má nejmenší rozdíl mezi minimální a maximální délkou křídel?

# 1. Vytvoříme tabulku statistik
stats = penguins.groupby("species")["flipper_length_mm"].agg(["min", "max"])

# 2. Vytvoříme nový sloupec 'rozdil'
stats["rozdil"] = stats["max"] - stats["min"]

# 3. Seřadíme podle rozdílu od nejmenšího
print(stats.sort_values("rozdil"))
```

```python
# Jaký je medián hmotnosti tučňáků a liší se tento medián mezi jednotlivými ostrovy, kde tučňáci žijí? Na kterém ostrově je medián hmotnosti největší?
# Data seskupte podle názvu ostrova a použijte agregační funkci median() pro výpočet mediánu hmotnosti tučňáků na každém ostrově.

median = penguins.groupby("island")["body_mass_g"].median()
median
```

```python
# Které druhy tučňáků mají průměrnou délku křídel větší než 200 mm?
# 1. Seskupit a spočítat průměr
avg_kridla = penguins.groupby("species")["flipper_length_mm"].mean()

# 2. Zobrazit jen ty nad 200 mm
(avg_kridla[avg_kridla > 200])
```
