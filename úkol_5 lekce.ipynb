```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/allisonhorst/palmerpenguins/master/inst/extdata/penguins.csv")
data
```

```python
# 1. Vytvoření grafu z dat a jeho uložení do proměnné 'ax'
ax = sns.countplot(data=data, x="species", palette="Set2")

# 2. Nastavení popisků pro tento konkrétní graf
ax.set(xlabel="vodorovna", ylabel="svisla", title="Graf")
```

```python
# Vlož kód pro vytvoření histogramu s tučnáky (včetně parametrů `bin`)
ax=sns.histplot(data=data, x="body_mass_g", bins=15)
```

```python
# Vlož kód pro vytvoření boxplotu s tučnáky:
graf = sns.boxplot(data=data, x="species", y="body_mass_g")
```
