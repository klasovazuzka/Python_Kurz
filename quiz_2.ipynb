```python
import pandas as pd
```

```python
data = pd.read_csv("https://raw.githubusercontent.com/allisonhorst/palmerpenguins/master/inst/extdata/penguins.csv")
```

```python
data.describe()
```

```python
#data.info()
data ["species"].unique()
druhy_dream = data[data['island'] == 'Dream']['species'].unique()
print(druhy_dream)

tezci_tucnaci = data[data['body_mass_g'] > 5000] # Filtrování podle hmotnosti

data_clean = data.dropna() # Vytvoření očištěné kopie bez prázdných řádků

vysledek = len(data_clean[data_clean['body_mass_g'] > 5000]) # A následně filtrovat na těchto datech
print(vysledek)

# Filtrace se dvěma podmínkami
filtrovani = data[(data['body_mass_g'] < 3500) & (data['flipper_length_mm'] > 200)]

# Zjištění počtu
vysledek = len(filtrovani)
print(f"Počet tučňáků: {vysledek}")

# Filtrace: ostrov je Torgersen A ZÁROVEŇ pohlaví je female
samice_torgersen = data[(data['island'] == 'Torgersen') & (data['sex'] == 'female')]

# Zjištění počtu
vysledek = len(samice_torgersen)
print(f"Počet samic na ostrově Torgersen: {vysledek}")

# Spočítáme výskyty pro jednotlivé roky
pocty_roku = data['year'].value_counts()
print(pocty_roku)

# Zobrazení statistik pro sloupec hmotnosti
statistiky = data['body_mass_g'].describe()
print(statistiky)

# Nebo přímo medián
median = data['body_mass_g'].median()
print(f"Medián: {median}")
```
